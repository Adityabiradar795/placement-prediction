```python
plt.figure(figsize=(12,8))
def box_plot(col,num):
    plt.subplot(4,4,num)
    sns.boxplot(data=df,y=col)
i=1
for col in num_col:
    box_plot(col,i)
    i+=1
plt.tight_layout()
plt.show()

# 'Attendance_Percentage', 'Study_Hours_Per_Week', 'CGPA', 'Age' apply iqr on these
```

```python
plt.figure(figsize=(12,8))
def box_plot(col,num):
    plt.subplot(4,4,num)
    sns.histplot(df[col],kde=True)
i=1
for col in num_col:
    box_plot(col,i)
    i+=1
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    fmt=".2f",
    cmap="rocket",
    annot_kws={"size": 8}
)

plt.xticks(rotation=60, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
def get_iqr_bounds(series,k=1.5):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    
    IQR = Q3-Q1
    
    lb = Q1 - k*IQR
    ub = Q3 + k*IQR
    return lb,ub

iqr_req_col = ['Attendance_Percentage', 'Study_Hours_Per_Week', 'CGPA', 'Age']
for col in iqr_req_col:
    lower,upper = get_iqr_bounds(df[col])
    df[col] = df[col].clip(lower = lower,upper = upper)
```
