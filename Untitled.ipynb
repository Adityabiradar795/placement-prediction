```python
def get_iqr_bounds(series,k=1.5):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    
    IQR = Q3-Q1
    
    lb = Q1 - k*IQR
    ub = Q3 + k*IQR
    return lb,ub

iqr_req_col = ['Attendance_Percentage', 'Study_Hours_Per_Week', 'CGPA', 'Age']
for col in iqr_req_col:
    lower,upper = get_iqr_bounds(df[col])
    df[col] = df[col].clip(lower = lower,upper = upper)
```

```python
placed_mask = df['Placement_Status'] == 'Placed'
placed_mask
```

```python
lower,upper = get_iqr_bounds(df.loc[placed_mask,'Starting_Salary_USD'])

df.loc[placed_mask,'Starting_Salary_USD'] = df.loc[placed_mask,'Starting_Salary_USD'].clip(lower=lower,upper=upper)
```
